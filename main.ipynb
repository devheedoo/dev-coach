```python
import os
import pprint
import warnings
from pathlib import Path

from dotenv import load_dotenv
from langgraph.checkpoint.memory import MemorySaver

from dev_coach import build_graph

load_dotenv()

warnings.filterwarnings(
    "ignore",
    message=r"The default value of `allowed_objects` will change",
    category=DeprecationWarning,
)

# Notebook 전용 SQLite 경로 (데모 데이터가 dev_coach.sqlite 와 섞이지 않도록)
os.environ.setdefault(
    "DEV_COACH_SQLITE_PATH",
    str(Path.cwd().resolve() / "_notebook_dev_coach.sqlite"),
)
os.environ.setdefault("DEV_COACH_LEARNER_ID", "notebook_demo")

graph = build_graph(checkpointer=MemorySaver())

THREAD_ID = "notebook-thread-1"
config = {"configurable": {"thread_id": THREAD_ID}}

initial_input = {
    "topic": "React performance",
    "learner_identifier": "notebook_demo",
    "profile": {
        "years_experience": 5,
        "role": "Frontend Engineer",
        "level": "Senior",
    },
}

step1 = graph.invoke(initial_input, config=config)

print("--- interrupt (waiting for answer) ---")
pprint.pp(step1.get("__interrupt__"))
print("\ncurrent_question:\n", step1.get("current_question"))
```
